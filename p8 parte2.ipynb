```python
import numpy as np
import cv2
import matplotlib.pyplot as plt


# CARGAR IMAGEN EN ESCALA DE GRISES


img = cv2.imread('rmcraneo.jpg', cv2.IMREAD_GRAYSCALE)

if img is None:
    print("No se pudo cargar la imagen")
    exit()

# TRANSFORMADA DE FOURIER
f = np.fft.fft2(img)

# Centrar frecuencias bajas
fshift = np.fft.fftshift(f)

# Espectro de magnitud
magnitude_spectrum = 20 * np.log(np.abs(fshift) + 1)

# Mostrar imagen original y espectro
plt.figure(figsize=(12,6))

plt.subplot(1,2,1)
plt.imshow(img, cmap='gray')
plt.title('Imagen Original')
plt.axis('off')

plt.subplot(1,2,2)
plt.imshow(magnitude_spectrum, cmap='gray')
plt.title('Espectro de Magnitud')
plt.axis('off')

plt.show()
# VARIABLES GENERALES

rows, cols = img.shape

crow, ccol = rows // 2, cols // 2

# FILTRO PASA BAJAS - RADIO 20
r1 = 20

mask_pb20 = np.zeros((rows, cols), np.uint8)

cv2.circle(mask_pb20, (ccol, crow), r1, 1, thickness=-1)

# Aplicar máscara
fshift_pb20 = fshift * mask_pb20

# Transformada inversa
f_ishift_pb20 = np.fft.ifftshift(fshift_pb20)

img_pb20 = np.fft.ifft2(f_ishift_pb20)

img_pb20 = np.abs(img_pb20)

# Mostrar resultados
plt.figure(figsize=(12,6))

plt.subplot(1,2,1)
plt.imshow(mask_pb20, cmap='gray')
plt.title('Mascara Pasa Bajas R=20')

plt.subplot(1,2,2)
plt.imshow(img_pb20, cmap='gray')
plt.title('Imagen Filtrada PB 20')

plt.show()
# FILTRO PASA BAJAS - RADIO 80
r2 = 80

mask_pb80 = np.zeros((rows, cols), np.uint8)

cv2.circle(mask_pb80, (ccol, crow), r2, 1, thickness=-1)

# Aplicar máscara
fshift_pb80 = fshift * mask_pb80

# Transformada inversa
f_ishift_pb80 = np.fft.ifftshift(fshift_pb80)

img_pb80 = np.fft.ifft2(f_ishift_pb80)

img_pb80 = np.abs(img_pb80)

# Mostrar resultados
plt.figure(figsize=(12,6))

plt.subplot(1,2,1)
plt.imshow(mask_pb80, cmap='gray')
plt.title('Mascara Pasa Bajas R=80')

plt.subplot(1,2,2)
plt.imshow(img_pb80, cmap='gray')
plt.title('Imagen Filtrada PB 80')

plt.show()

# FILTRO PASA ALTAS CIRCULAR

mask_pa_circle = np.ones((rows, cols), np.uint8)

cv2.circle(mask_pa_circle, (ccol, crow), 30, 0, thickness=-1)

# Aplicar máscara
fshift_pa_circle = fshift * mask_pa_circle

# Transformada inversa
f_ishift_pa_circle = np.fft.ifftshift(fshift_pa_circle)

img_pa_circle = np.fft.ifft2(f_ishift_pa_circle)

img_pa_circle = np.abs(img_pa_circle)

# Mostrar resultados
plt.figure(figsize=(12,6))

plt.subplot(1,2,1)
plt.imshow(mask_pa_circle, cmap='gray')
plt.title('Mascara Pasa Altas Circular')

plt.subplot(1,2,2)
plt.imshow(img_pa_circle, cmap='gray')
plt.title('Imagen PA Circular')

plt.show()

# FILTRO PASA ALTAS EN CRUZ

mask_pa_cross = np.ones((rows, cols), np.uint8)

mask_pa_cross[crow-10:crow+10, :] = 0
mask_pa_cross[:, ccol-10:ccol+10] = 0

# Aplicar máscara
fshift_pa_cross = fshift * mask_pa_cross

# Transformada inversa
f_ishift_pa_cross = np.fft.ifftshift(fshift_pa_cross)

img_pa_cross = np.fft.ifft2(f_ishift_pa_cross)

img_pa_cross = np.abs(img_pa_cross)

# Mostrar resultados
plt.figure(figsize=(12,6))

plt.subplot(1,2,1)
plt.imshow(mask_pa_cross, cmap='gray')
plt.title('Mascara Pasa Altas Cruz')

plt.subplot(1,2,2)
plt.imshow(img_pa_cross, cmap='gray')
plt.title('Imagen PA Cruz')

plt.show()
```
